## 6. Series and DataFrames

A **Series** is a labeled 1D array. A **DataFrame** is a 2D table of Series.

````python
import pandas as pd
import numpy as np

# Series
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
print(f"Series:\n{s}")
print(f"\ns['b']: {s['b']}")
print(f"s.mean(): {s.mean()}")
````

````python
# DataFrame from a dict
data = {
    "name": ["Ada", "Bob", "Eve", "Max"],
    "age": [36, 25, 30, 42],
    "city": ["London", "NYC", "Paris", "Berlin"],
    "salary": [120000, 80000, 95000, 110000],
}
df = pd.DataFrame(data)
print(df)
````

## 7. Inspecting Data

Quick ways to understand a DataFrame.

````python
# Inspecting data
print("head():")
print(df.head(2))

print("\ninfo():")
print(df.info())

print("\ndescribe():")
print(df.describe())

print(f"\nshape: {df.shape}")
print(f"columns: {list(df.columns)}")
print(f"dtypes:\n{df.dtypes}")
````

## 8. Selecting Data

Select columns and rows using labels or positions.

````python
# Selecting columns
print("df['name']:")
print(df["name"])

print("\ndf[['name', 'age']]:")
print(df[["name", "age"]])
````

````python
# Selecting rows with .loc (label) and .iloc (position)
print("df.loc[0]:")
print(df.loc[0])

print("\ndf.iloc[1:3]:")
print(df.iloc[1:3])

print("\ndf.loc[df['age'] > 30]:")
print(df.loc[df["age"] > 30])
````

## 9. Filtering

Filter rows based on conditions.

````python
# Filtering
print("People over 30:")
print(df[df["age"] > 30])

print("\nPeople in London or Paris:")
print(df[df["city"].isin(["London", "Paris"])])

print("\nPeople with salary > 100k and age < 40:")
print(df[(df["salary"] > 100000) & (df["age"] < 40)])
````

## 10. Handling Missing Values

Real data has missing values. Pandas uses `NaN` (Not a Number).

````python
# Missing values
df_missing = pd.DataFrame({
    "name": ["Ada", "Bob", "Eve"],
    "age": [36, None, 30],
    "salary": [120000, 80000, None],
})
print("df_missing:")
print(df_missing)

print("\nisnull():")
print(df_missing.isnull())

print("\nDrop rows with any missing:")
print(df_missing.dropna())

print("\nFill missing with a value:")
print(df_missing.fillna({"age": 0, "salary": 0}))
````

## 11. Grouping and Aggregation

Group data by a column and compute statistics per group.

````python
# Grouping and aggregation
sales = pd.DataFrame({
    "region": ["North", "South", "North", "South", "North"],
    "product": ["A", "A", "B", "B", "A"],
    "revenue": [100, 150, 200, 120, 180],
})
print("sales:")
print(sales)

print("\nRevenue by region:")
print(sales.groupby("region")["revenue"].sum())

print("\nMean revenue by region:")
print(sales.groupby("region")["revenue"].mean())

print("\nMultiple stats:")
print(sales.groupby("region")["revenue"].agg(["sum", "mean", "count"]))
````

## 12. Adding and Removing Columns

Create new columns from existing ones.

````python
# Adding and removing columns
df2 = df.copy()

# Add a column
df2["age_in_months"] = df2["age"] * 12
print("After adding age_in_months:")
print(df2)

# Add a column based on condition
df2["senior"] = df2["age"] >= 35
print("\nAfter adding senior:")
print(df2)

# Drop a column
df2 = df2.drop(columns=["age_in_months"])
print("\nAfter dropping age_in_months:")
print(df2.columns)
````

## 13. Merging DataFrames

Combine DataFrames like SQL joins.

````python
# Merging DataFrames
employees = pd.DataFrame({
    "emp_id": [1, 2, 3],
    "name": ["Ada", "Bob", "Eve"],
})
departments = pd.DataFrame({
    "emp_id": [1, 2, 4],
    "dept": ["AI", "Data", "Sales"],
})

print("employees:")
print(employees)
print("\ndepartments:")
print(departments)

# Inner join (only matching emp_id)
merged = pd.merge(employees, departments, on="emp_id", how="inner")
print("\nInner join:")
print(merged)

# Left join (keep all employees)
left = pd.merge(employees, departments, on="emp_id", how="left")
print("\nLeft join:")
print(left)
````

## 15. Experiment: Analyzing a Dataset

Let's create a small dataset and analyze it end-to-end.

````python
# Experiment: analyze a dataset
np.random.seed(42)
n = 100
students = pd.DataFrame({
    "student_id": range(1, n + 1),
    "study_hours": np.random.uniform(0, 10, n).round(1),
    "attendance": np.random.uniform(50, 100, n).round(1),
    "passed": np.random.choice([0, 1], n, p=[0.3, 0.7]),
})

print("First 5 rows:")
print(students.head())

print("\nSummary:")
print(students.describe())

print("\nPass rate:")
print(students["passed"].mean())

print("\nAverage study hours by pass status:")
print(students.groupby("passed")["study_hours"].mean())

print("\nStudents who studied > 5 hours and passed:")
print(students[(students["study_hours"] > 5) & (students["passed"] == 1)].shape[0], "students")
````

## 16. Failure Case: Chained Assignment

A common Pandas pitfall is **chained assignment** — modifying a slice of a slice, which may not work as expected.

````python
# Chained assignment pitfall
df3 = df.copy()

# BAD: chained assignment (may not modify original)
df3[df3["age"] > 30]["salary"] = 0  # SettingWithCopyWarning

# GOOD: use .loc
df3.loc[df3["age"] > 30, "salary"] = 0
print("After setting salary=0 for age>30 using .loc:")
print(df3)
````

## 17. Debugging: Common Errors

### KeyError — column doesn't exist
```python
df["nonexistent"]  # KeyError
```

### SettingWithCopyWarning
Chained assignment. Fix: use `.loc`.

### NaN propagation
Operations with NaN produce NaN. Use `dropna()` or `fillna()`.

### dtype issues
Mixed types in a column can cause unexpected behavior.

## 18. Real-World Considerations

- **Memory**: large DataFrames use lots of memory. Use appropriate dtypes (`category`, `float32`).
- **Vectorized ops**: avoid `apply` with Python loops when vectorized ops work.
- **Copy vs view**: use `.copy()` when you'll modify a slice.
- **Index**: reset/set index deliberately.
- **Pipeline**: chain operations with `.pipe()` or method chaining for readability.

## 19. Common Mistakes

- Chained assignment
- Forgetting to handle NaN
- Using `apply` instead of vectorized ops
- Modifying a view thinking it's a copy
- Confusing `.loc` and `.iloc`

## 20. When NOT to Use

- Don't use Pandas for large numeric arrays — use NumPy.
- Don't use Pandas for deep learning tensors — use PyTorch.
- Don't use Pandas for very large data that doesn't fit in memory — use Dask/Spark.

## 21. Challenge

Given the `students` DataFrame, compute:
1. The average attendance of students who passed.
2. The number of students who studied less than 3 hours.
3. A new column `performance` = `study_hours * attendance / 100`.

````python
# Challenge
avg_att_pass = students[students["passed"] == 1]["attendance"].mean()
print(f"Average attendance of passers: {avg_att_pass:.2f}")

low_study = (students["study_hours"] < 3).sum()
print(f"Students who studied < 3 hours: {low_study}")

students["performance"] = students["study_hours"] * students["attendance"] / 100
print("\nWith performance column:")
print(students.head())
````
