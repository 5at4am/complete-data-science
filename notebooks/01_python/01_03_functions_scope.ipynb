## What Are We Solving?

Repeating the same logic over and over is error-prone and unmaintainable. If you copy-paste code, every bug fix must be applied everywhere. **Functions** let you package logic into reusable, named blocks.

**Scope** defines where variables are visible. Without understanding scope, you'll encounter mysterious bugs where variables seem to disappear or have unexpected values.

## Why Does It Matter?

Every ML library is a collection of functions:
- `model.fit(X, y)` — a function that trains a model
- `model.predict(X)` — a function that makes predictions
- `np.array([...])` — a function that creates an array

Understanding how to write, call, and compose functions — and how variable scope works — is essential for writing clean, debuggable ML code.

## Mental Model

A function is a **machine**: you feed it inputs (parameters), it does work, and it produces outputs (return values).

```
Input (parameters)  →  [Function Body]  →  Output (return value)
                     (does work here)
```

Scope is like **rooms in a house** — variables in one room aren't visible in another unless explicitly shared.

## Defining and Calling Functions

Use `def` to define a function. Call it with parentheses.

```python
def function_name(parameters):
    """Docstring: explain what this function does."""
    # function body
    return result  # optional
```

**Rules:**
- Function names follow `snake_case` convention
- Always include a docstring (explain *what* and *why*)
- Use `return` to send a value back; without it, the function returns `None`

````python
# Define a function
def greet(name):
    return f"Hello, {name}!"

# Call it
message = greet("Ada")
print(message)

# Function with no parameters
def say_hello():
    print("Hello!")

say_hello()
````

### Parameters, Arguments, and Default Values

| Term | Meaning | Example |
|------|---------|---------|
| **Parameter** | Variable in the function definition | `def greet(name):` — `name` is a parameter |
| **Argument** | Value passed when calling | `greet("Ada")` — `"Ada"` is an argument |
| **Default value** | Value used when argument not provided | `def greet(name="World"):` |

**Order matters:** Required parameters first, then optional (default) parameters.

````python
# Function with multiple parameters and default value
def power(base, exponent=2):
    return base ** exponent

print(f"power(3): {power(3)}")          # uses default exponent=2
print(f"power(3, 3): {power(3, 3)}")    # positional
print(f"power(base=2, exponent=5): {power(base=2, exponent=5)}")  # keyword
````

### Returning Multiple Values

Python functions can return multiple values as a tuple. This is very common in ML — many functions return (value, metadata) pairs.

````python
# Return multiple values
def min_max(numbers):
    return min(numbers), max(numbers)

lo, hi = min_max([3, 1, 4, 1, 5, 9, 2])
print(f"min: {lo}, max: {hi}")
````

## `*args` and `**kwargs`

These let you write flexible functions that accept any number of arguments.

| Syntax | What it collects | Type | Example |
|--------|-----------------|------|---------|
| `*args` | Extra positional args | Tuple | `f(1, 2, 3)` → `args = (1, 2, 3)` |
| `**kwargs` | Extra keyword args | Dict | `f(a=1, b=2)` → `kwargs = {'a': 1, 'b': 2}` |

**ML relevance:** Most ML functions use `**kwargs` for optional parameters like `verbose=True`, `random_state=42`, etc.

````python
# *args: variable number of positional args
def total(*args):
    return sum(args)

print(f"total(1, 2, 3): {total(1, 2, 3)}")
print(f"total(1, 2, 3, 4, 5): {total(1, 2, 3, 4, 5)}")
````

````python
# **kwargs: variable number of keyword args
def describe(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

describe(name="Ada", age=36, city="London")
````

## Scope

**Scope** defines where a variable is accessible. Python uses the **LEGB** rule:

| Scope | Where | When to use |
|-------|-------|-------------|
| **L**ocal | Inside the function | Default for variables created in a function |
| **E**nclosing | In an outer function | For closures |
| **G**lobal | At the module level | Use sparingly, prefer passing as arguments |
| **B**uilt-in | Python's builtins | `print`, `len`, `range` — don't override these |

**Rule of thumb:** If you need `global`, your function is probably doing too much. Pass values as arguments instead.

````python
# Global vs local scope
x = 10  # global

def my_func():
    y = 5  # local
    print(f"Inside function: x={x} (global), y={y} (local)")

my_func()
print(f"Outside function: x={x} (global)")

# y is not accessible outside the function
# print(y)  # NameError: name 'y' is not defined
````

````python
# Modifying a global inside a function requires 'global'
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
increment()
print(f"counter after 3 increments: {counter}")
````

## Lambda Functions

A **lambda** is a small, anonymous, single-expression function. Useful for short, throwaway logic.

```python
# lambda is equivalent to:
square = lambda x: x**2
# is the same as:
def square(x):
    return x**2
```

**When to use:**
- As a key for `sorted()` or `max()`
- With `map()` and `filter()`
- **Don't use** for complex logic — use a named function instead

````python
# Lambda functions
square = lambda x: x**2
print(f"square(5): {square(5)}")

# Use with sorted
people = [("Ada", 36), ("Bob", 25), ("Eve", 30)]
by_age = sorted(people, key=lambda p: p[1])
print(f"sorted by age: {by_age}")

# Use with map
nums = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, nums))
print(f"doubled: {doubled}")
````

## Closures

A **closure** is a function that remembers variables from its enclosing scope even after the outer function has finished.

```python
def make_multiplier(factor):
    def multiply(x):
        return x * factor  # 'factor' is remembered!
    return multiply

double = make_multiplier(2)  # factor=2 is "closed over"
triple = make_multiplier(3)  # factor=3 is "closed over"
```

**ML relevance:** Closures are used in decorators (for timing, logging) and in creating factory functions for model components.

````python
# Closure: a function that remembers its enclosing scope
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(f"double(5): {double(5)}")
print(f"triple(5): {triple(5)}")
print(f"double remembers factor=2, triple remembers factor=3")
````

## Experiment: A Simple Calculator

Build a small calculator using functions. This demonstrates how functions compose — small functions combine into larger capabilities.

````python
# Calculator using functions
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b):
    if b == 0:
        return "Error: division by zero"
    return a / b

operations = {
    "+": add,
    "-": subtract,
    "*": multiply,
    "/": divide,
}

def calculate(a, op, b):
    func = operations.get(op)
    if func is None:
        return f"Unknown operator: {op}"
    return func(a, b)

print(f"10 + 5 = {calculate(10, '+', 5)}")
print(f"10 - 5 = {calculate(10, '-', 5)}")
print(f"10 * 5 = {calculate(10, '*', 5)}")
print(f"10 / 5 = {calculate(10, '/', 5)}")
print(f"10 / 0 = {calculate(10, '/', 0)}")
print(f"10 % 5 = {calculate(10, '%', 5)}")
````

## Common Mistakes

| Mistake | What goes wrong | How to fix it |
|---------|----------------|---------------|
| Mutable default argument | `def f(x, lst=[])` — shared across calls | Use `lst=None` then `if lst is None: lst = []` |
| Overusing `global` | Hard to debug, side effects everywhere | Pass values as arguments, return results |
| Functions that do too much | Hard to test, hard to reuse | Single responsibility: one function = one job |
| Confusing `return` with `print` | `print` outputs to console, doesn't return data | Use `return` when you need the value elsewhere |
| Forgetting `self` in methods | `TypeError: missing positional argument` | Always add `self` as the first parameter |
| No docstring | Others (and future you) can't understand the function | Always add a docstring explaining what/why |

````python
# Recap: mutable default argument bug
def append_to(element, target=[]):
    target.append(element)
    return target

print(append_to(1))  # [1]
print(append_to(2))  # [1, 2]  <- BUG: shared default

# Correct pattern
def append_to_fixed(element, target=None):
    if target is None:
        target = []
    target.append(element)
    return target

print(append_to_fixed(1))  # [1]
print(append_to_fixed(2))  # [2]  <- correct
````

````python
# Demonstrate UnboundLocalError and fix
x = 10
try:
    def f():
        print(x)
        x = 5
    f()
except UnboundLocalError as e:
    print(f"UnboundLocalError: {e}")
    print("Fix: use 'global x' or rename the local variable")

    def f_fixed():
        global x
        print(f"x before: {x}")
        x = 5
        print(f"x after: {x}")
    f_fixed()
````

## Real-World Considerations

- **Single responsibility**: each function should do one thing. If you can't describe it in one sentence, split it.
- **Docstrings**: document what a function does (`"""..."""`). Include parameters and return value.
- **Type hints**: annotate parameter and return types for clarity: `def add(a: int, b: int) -> int:`
- **Avoid `global`**: prefer passing values as arguments and returning results.
- **Pure functions**: functions that don't modify external state are easier to test and debug.

````python
# Challenge: Fibonacci
def fibonacci(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    seq = [0, 1]
    for i in range(2, n):
        seq.append(seq[-1] + seq[-2])
    return seq

print(f"fibonacci(10): {fibonacci(10)}")
````
