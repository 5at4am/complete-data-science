## 5. Mental Model

A function is a **machine**: you feed it inputs (parameters), it does work, and it produces outputs (return values). Scope is like **rooms in a house** — variables in one room aren't visible in another unless explicitly shared.

## 6. Defining and Calling Functions

Use `def` to define a function. Call it with parentheses.

````python
# Define a function
def greet(name):
    return f"Hello, {name}!"

# Call it
message = greet("Ada")
print(message)

# Function with no parameters
def say_hello():
    print("Hello!")

say_hello()
````

````python
# Function with multiple parameters and default value
def power(base, exponent=2):
    return base ** exponent

print(f"power(3): {power(3)}")          # uses default exponent=2
print(f"power(3, 3): {power(3, 3)}")    # positional
print(f"power(base=2, exponent=5): {power(base=2, exponent=5)}")  # keyword
````

## 7. Returning Multiple Values

Python functions can return multiple values as a tuple.

````python
# Return multiple values
def min_max(numbers):
    return min(numbers), max(numbers)

lo, hi = min_max([3, 1, 4, 1, 5, 9, 2])
print(f"min: {lo}, max: {hi}")
````

## 8. `*args` and `**kwargs`

- `*args` — collect extra positional arguments into a tuple
- `**kwargs` — collect extra keyword arguments into a dict

````python
# *args: variable number of positional args
def total(*args):
    return sum(args)

print(f"total(1, 2, 3): {total(1, 2, 3)}")
print(f"total(1, 2, 3, 4, 5): {total(1, 2, 3, 4, 5)}")
````

````python
# **kwargs: variable number of keyword args
def describe(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

describe(name="Ada", age=36, city="London")
````

## 9. Scope

**Scope** defines where a variable is accessible.

- **Local** — inside a function
- **Global** — at the module level
- **Enclosing** — in an outer function (for closures)
- **Built-in** — Python's built-in names

````python
# Global vs local scope
x = 10  # global

def my_func():
    y = 5  # local
    print(f"Inside function: x={x} (global), y={y} (local)")

my_func()
print(f"Outside function: x={x} (global)")

# y is not accessible outside the function
# print(y)  # NameError: name 'y' is not defined
````

````python
# Modifying a global inside a function requires 'global'
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
increment()
print(f"counter after 3 increments: {counter}")
````

## 10. Lambda Functions

A **lambda** is a small, anonymous, single-expression function. Useful for short, throwaway logic.

````python
# Lambda functions
square = lambda x: x**2
print(f"square(5): {square(5)}")

# Use with sorted
people = [("Ada", 36), ("Bob", 25), ("Eve", 30)]
by_age = sorted(people, key=lambda p: p[1])
print(f"sorted by age: {by_age}")

# Use with map
nums = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, nums))
print(f"doubled: {doubled}")
````

## 11. Closures

A **closure** is a function that remembers variables from its enclosing scope even after the outer function has finished.

````python
# Closure: a function that remembers its enclosing scope
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(f"double(5): {double(5)}")
print(f"triple(5): {triple(5)}")
print(f"double remembers factor=2, triple remembers factor=3")
````

## 12. Experiment: A Simple Calculator

Build a small calculator using functions.

````python
# Calculator using functions
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b):
    if b == 0:
        return "Error: division by zero"
    return a / b

operations = {
    "+": add,
    "-": subtract,
    "*": multiply,
    "/": divide,
}

def calculate(a, op, b):
    func = operations.get(op)
    if func is None:
        return f"Unknown operator: {op}"
    return func(a, b)

print(f"10 + 5 = {calculate(10, '+', 5)}")
print(f"10 - 5 = {calculate(10, '-', 5)}")
print(f"10 * 5 = {calculate(10, '*', 5)}")
print(f"10 / 5 = {calculate(10, '/', 5)}")
print(f"10 / 0 = {calculate(10, '/', 0)}")
print(f"10 % 5 = {calculate(10, '%', 5)}")
````

## 13. Failure Case: Mutable Default Argument (Recap)

We saw this in Unit 01.2. It's a function-scope issue: the default list is created once and shared.

````python
# Recap: mutable default argument bug
def append_to(element, target=[]):
    target.append(element)
    return target

print(append_to(1))  # [1]
print(append_to(2))  # [1, 2]  <- BUG: shared default

# Correct pattern
def append_to_fixed(element, target=None):
    if target is None:
        target = []
    target.append(element)
    return target

print(append_to_fixed(1))  # [1]
print(append_to_fixed(2))  # [2]  <- correct
````

## 14. Debugging: Common Errors

### NameError — variable not in scope
```python
def f():
    print(x)  # x not defined in this scope
```

### UnboundLocalError — local used before assignment
```python
x = 10
def f():
    print(x)  # UnboundLocalError
    x = 5
```
Because `x` is assigned in the function, Python treats it as local throughout.

### TypeError — wrong number of arguments
```python
def f(a, b): pass
f(1)  # TypeError: missing 1 required positional argument
```

````python
# Demonstrate UnboundLocalError and fix
x = 10
try:
    def f():
        print(x)
        x = 5
    f()
except UnboundLocalError as e:
    print(f"UnboundLocalError: {e}")
    print("Fix: use 'global x' or rename the local variable")

    def f_fixed():
        global x
        print(f"x before: {x}")
        x = 5
        print(f"x after: {x}")
    f_fixed()
````

## 15. Real-World Considerations

- **Single responsibility**: each function should do one thing.
- **Docstrings**: document what a function does (`"""..."""`).
- **Type hints**: annotate parameter and return types for clarity.
- **Avoid `global`**: prefer passing values as arguments and returning results.
- **Pure functions**: functions that don't modify external state are easier to test and debug.

## 16. Common Mistakes

- Mutable default arguments
- Overusing `global`
- Functions that are too long / do too much
- Not returning values when needed
- Confusing `return` with `print`

## 17. When NOT to Use

- Don't use a lambda for complex logic — use a named function.
- Don't use `global` when you can pass arguments.

## 18. Challenge

Write a function `fibonacci(n)` that returns the first `n` Fibonacci numbers using a loop.

````python
# Challenge: Fibonacci
def fibonacci(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    seq = [0, 1]
    for i in range(2, n):
        seq.append(seq[-1] + seq[-2])
    return seq

print(f"fibonacci(10): {fibonacci(10)}")
````
