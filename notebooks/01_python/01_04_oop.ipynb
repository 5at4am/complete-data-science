## 6. Defining a Class

Use `class` to define a class. `__init__` is the constructor that runs when you create an object. `self` refers to the current instance.

````python
# Define a class
class Dog:
    def __init__(self, name, age):
        self.name = name   # instance attribute
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"

    def birthday(self):
        self.age += 1
        return f"{self.name} is now {self.age} years old"

# Create objects (instances)
rex = Dog("Rex", 3)
bella = Dog("Bella", 5)

print(rex.bark())
print(bella.bark())
print(rex.birthday())
print(f"Rex's age: {rex.age}")
````

## 7. Instance vs Class Attributes

- **Instance attributes** — set in `__init__`, unique per object.
- **Class attributes** — shared across all instances.

````python
# Class vs instance attributes
class Animal:
    kingdom = "Animalia"  # class attribute (shared)

    def __init__(self, species):
        self.species = species  # instance attribute

a1 = Animal("Cat")
a2 = Animal("Dog")

print(f"a1.kingdom: {a1.kingdom}")
print(f"a2.kingdom: {a2.kingdom}")
print(f"a1.species: {a1.species}")
print(f"a2.species: {a2.species}")

# Class attribute is shared
Animal.kingdom = "Animalia (updated)"
print(f"a1.kingdom after update: {a1.kingdom}")
````

## 8. Class Methods and Static Methods

- **`@classmethod`** — takes `cls`, operates on the class.
- **`@staticmethod`** — takes no `self`/`cls`, just a utility function in the class.

````python
# Class methods and static methods
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @classmethod
    def from_fahrenheit(cls, fahrenheit):
        celsius = (fahrenheit - 32) * 5 / 9
        return cls(celsius)

    @staticmethod
    def is_freezing(celsius):
        return celsius <= 0

t1 = Temperature(25)
t2 = Temperature.from_fahrenheit(32)  # 0 C

print(f"t1.celsius: {t1.celsius}")
print(f"t2.celsius (from 32F): {t2.celsius}")
print(f"Is 0C freezing? {Temperature.is_freezing(0)}")
print(f"Is 10C freezing? {Temperature.is_freezing(10)}")
````

## 9. Inheritance

**Inheritance** lets a class (child) reuse and extend another class (parent). The child inherits attributes and methods, and can override them.

````python
# Inheritance
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        return f"{self.brand} vehicle starting..."

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)  # call parent constructor
        self.doors = doors

    def start(self):
        # override parent method
        return f"{self.brand} car with {self.doors} doors starting..."

    def honk(self):
        return "Beep beep!"

car = Car("Toyota", 4)
print(car.start())   # overridden
print(car.honk())    # child-specific
print(f"brand: {car.brand}, doors: {car.doors}")
````

## 10. Special (Dunder) Methods

Dunder (double-underscore) methods define how objects behave with Python operators and built-ins.

````python
# Special methods
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __len__(self):
        return 2

v1 = Vector(1, 2)
v2 = Vector(3, 4)

print(f"v1: {v1}")                    # __repr__
print(f"v1 + v2: {v1 + v2}")          # __add__
print(f"v1 == Vector(1,2): {v1 == Vector(1, 2)}")  # __eq__
print(f"len(v1): {len(v1)}")          # __len__
````

## 11. Encapsulation

**Encapsulation** hides internal state and exposes controlled access. Python uses naming conventions:

- `_name` — protected (convention: internal use)
- `__name` — name-mangled (stronger privacy)

````python
# Encapsulation with properties
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance  # protected

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

acc = BankAccount("Ada", 100)
print(f"balance: {acc.balance}")
acc.deposit(50)
print(f"after deposit: {acc.balance}")
acc.withdraw(30)
print(f"after withdraw: {acc.balance}")

try:
    acc.withdraw(1000)
except ValueError as e:
    print(f"Error: {e}")
````

## 12. Experiment: A Simple ML-Style Model Class

Let's build a class that mimics the structure of a scikit-learn estimator (fit/predict).

````python
# A simple linear model class (fit/predict pattern)
class SimpleLinearModel:
    def __init__(self, learning_rate=0.01, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.slope = 0.0
        self.intercept = 0.0
        self.fitted = False

    def fit(self, X, y):
        # Gradient descent for y = slope*x + intercept
        n = len(X)
        for _ in range(self.epochs):
            pred = [self.slope * x + self.intercept for x in X]
            error = [p - t for p, t in zip(pred, y)]
            d_slope = (2 / n) * sum(e * x for e, x in zip(error, X))
            d_intercept = (2 / n) * sum(error)
            self.slope -= self.learning_rate * d_slope
            self.intercept -= self.learning_rate * d_intercept
        self.fitted = True
        return self

    def predict(self, X):
        if not self.fitted:
            raise RuntimeError("Model not fitted yet. Call fit() first.")
        return [self.slope * x + self.intercept for x in X]

# Generate data: y = 2x + 3
X = [i for i in range(10)]
y = [2 * i + 3 for i in range(10)]

model = SimpleLinearModel(learning_rate=0.01, epochs=1000)
model.fit(X, y)

print(f"Learned slope: {model.slope:.3f} (true: 2.0)")
print(f"Learned intercept: {model.intercept:.3f} (true: 3.0)")
print(f"predict(5): {model.predict([5])[0]:.2f} (expected: 13.0)")
````

## 13. Failure Case: Forgetting `self`

A common beginner bug is forgetting `self` as the first parameter of a method.

````python
# BUG: forgetting self
try:
    class Broken:
        def greet():  # missing self
            return "hi"
    b = Broken()
    print(b.greet())
except TypeError as e:
    print(f"TypeError: {e}")
    print("Fix: add self as the first parameter")

    class Fixed:
        def greet(self):
            return "hi"
    print(Fixed().greet())
````

## 14. Debugging: Common Errors

### AttributeError — attribute doesn't exist
```python
obj.missing_attr  # AttributeError
```

### TypeError — wrong number of args to method
```python
obj.method()  # if method requires self + args
```

### Forgetting `super().__init__()` in child
If the parent has required attributes, the child must call `super().__init__()`.

## 15. Real-World Considerations

- **Composition over inheritance**: prefer combining objects over deep inheritance chains.
- **Keep classes focused**: one class = one responsibility.
- **Use properties** for controlled attribute access.
- **Dunder methods** make your objects behave like built-ins.
- **Type hints** improve readability and tooling support.

## 16. Common Mistakes

- Forgetting `self`
- Forgetting `super().__init__()`
- Overusing inheritance (deep chains)
- Exposing mutable internal state directly
- Confusing class and instance attributes

## 17. When NOT to Use

- Don't use a class for a single function — use a function.
- Don't force OOP where simple functions suffice.

## 18. Challenge

Create a `Rectangle` class with `width` and `height`, and methods `area()` and `perimeter()`. Add a `__repr__` and a `__eq__`.

````python
# Challenge: Rectangle class
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def __repr__(self):
        return f"Rectangle({self.width}, {self.height})"

    def __eq__(self, other):
        return self.width == other.width and self.height == other.height

r1 = Rectangle(3, 4)
r2 = Rectangle(3, 4)
r3 = Rectangle(5, 2)

print(f"r1: {r1}")
print(f"r1.area(): {r1.area()}")
print(f"r1.perimeter(): {r1.perimeter()}")
print(f"r1 == r2: {r1 == r2}")
print(f"r1 == r3: {r1 == r3}")
````
