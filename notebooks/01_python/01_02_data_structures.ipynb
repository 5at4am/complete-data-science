## What Are We Solving?

Real data isn't a single number — it's collections of numbers, text, and objects. Python provides four core data structures to organize data: **lists**, **tuples**, **sets**, and **dictionaries**.

Without these, you'd be stuck storing one value at a time. That's like trying to do accounting with a single sticky note instead of a spreadsheet.

## Why Does It Matter?

Pandas DataFrames, NumPy arrays, and ML datasets are all built on these fundamentals:

- **Lists** → become columns in a DataFrame
- **Dicts** → become JSON payloads, config files, model parameters
- **Sets** → deduplicate categories, find common features
- **Tuples** → return multiple values from functions (very common in ML APIs)

Choosing the right structure (e.g., a set for fast membership checks, a dict for key-value lookup) is a core skill that separates efficient code from slow, buggy code.

## Mental Model

Think of each structure as a different **container**:

| Structure | Analogy | Key property | When to use |
|-----------|---------|-------------|-------------|
| **List** | Shopping list | Ordered, changeable | When order matters and you need to modify |
| **Tuple** | GPS coordinates | Ordered, unchangeable | When data must not change |
| **Set** | Bag of distinct marbles | Unordered, unique only | Deduplication, fast membership |
| **Dict** | Phone book | Key→value lookup | When you need to look things up by name |

## Lists

A **list** is an ordered, mutable (changeable) collection. It can hold any mix of types.

```python
# Lists use square brackets
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True]  # any types, even mixed
```

**Key insight:** Lists are the most commonly used data structure in Python. You'll use them constantly for batching data, storing results, and passing arguments.

````python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True]

print(f"fruits: {fruits}")
print(f"numbers: {numbers}")
print(f"mixed: {mixed}")
print(f"Length of fruits: {len(fruits)}")
````

### Indexing and Slicing

Python uses **0-based indexing** — the first element is at index 0.

```
Index:    0     1     2     3     4
Value:   10    20    30    40    50
Index:   -5    -4    -3    -2    -1   (negative = from end)
```

**Slicing syntax:** `list[start:stop:step]`
- `start` — inclusive (where to begin)
- `stop` — exclusive (where to end, but don't include it)
- `step` — optional, default 1

````python
# Indexing (0-based) and slicing
nums = [10, 20, 30, 40, 50]

print(f"nums[0]: {nums[0]}")       # first
print(f"nums[-1]: {nums[-1]}")     # last
print(f"nums[1:3]: {nums[1:3]}")   # slice [1,3)
print(f"nums[:2]: {nums[:2]}")     # first two
print(f"nums[::2]: {nums[::2]}")   # every other
print(f"nums[::-1]: {nums[::-1]}") # reversed
````

### List Methods

| Method | What it does | Mutates list? | Returns |
|--------|-------------|---------------|---------|
| `.append(x)` | Add x to end | Yes | None |
| `.insert(i, x)` | Insert x at index i | Yes | None |
| `.remove(x)` | Remove first occurrence of x | Yes | None |
| `.pop(i)` | Remove and return item at i | Yes | The removed item |
| `.sort()` | Sort in place | Yes | None |
| `.reverse()` | Reverse in place | Yes | None |
| `.index(x)` | Find index of x | No | Index |
| `.count(x)` | Count occurrences of x | No | Count |

**Tip:** `sorted(list)` returns a new sorted list without modifying the original. Use it when you want to preserve the original.

````python
# List methods
lst = [3, 1, 2]
lst.append(4)          # add to end
print(f"after append: {lst}")
lst.insert(0, 0)       # insert at index
print(f"after insert: {lst}")
lst.remove(1)          # remove first occurrence of value
print(f"after remove(1): {lst}")
popped = lst.pop()     # remove and return last
print(f"popped: {popped}, list: {lst}")
lst.sort()             # sort in place
print(f"after sort: {lst}")
lst.reverse()          # reverse in place
print(f"after reverse: {lst}")
````

## Tuples

A **tuple** is an ordered, **immutable** (unchangeable) collection. Once created, you can't modify it.

```python
# Tuples use parentheses
point = (3, 4)
rgb = (255, 0, 0)

# Or just comma-separated (parentheses optional)
point = 3, 4
```

**When to use tuples instead of lists:**
- Coordinates, RGB values, date components — data that shouldn't change
- Dictionary keys (lists can't be keys)
- Function return values (multiple returns)
- Unpacking: `x, y = point`

````python
# Tuples
point = (3, 4)
rgb = (255, 0, 0)

print(f"point: {point}")
print(f"point[0]: {point[0]}")
print(f"rgb: {rgb}")

# Tuple unpacking
x, y = point
print(f"Unpacked: x={x}, y={y}")

# Tuples are immutable - this would fail:
# point[0] = 5  # TypeError: 'tuple' object does not support item assignment
````

````python
# Demonstrate immutability
try:
    point = (3, 4)
    point[0] = 5
except TypeError as e:
    print(f"TypeError: {e}")
    print("Tuples cannot be modified after creation.")
````

## Sets

A **set** is an unordered collection of **unique** items. Great for removing duplicates and fast membership checks.

```python
# Sets use curly braces (but {} alone creates a dict!)
s = {1, 2, 3}
s = set()  # empty set (not {})

# Duplicates are automatically removed
s = {1, 2, 2, 3, 3, 3}  # → {1, 2, 3}
```

**Key insight:** Sets are **much faster** for membership checks than lists. If you have 1 million items and need to check if something exists, use a set (O(1)) not a list (O(n)).

````python
# Sets
s = {1, 2, 2, 3, 3, 3}   # duplicates removed automatically
print(f"set: {s}")

# Remove duplicates from a list
nums = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(nums))
print(f"original: {nums}")
print(f"unique: {unique}")
````

````python
# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"a: {a}")
print(f"b: {b}")
print(f"union (a | b): {a | b}")
print(f"intersection (a & b): {a & b}")
print(f"difference (a - b): {a - b}")
print(f"symmetric diff (a ^ b): {a ^ b}")

# Fast membership check
print(f"3 in a: {3 in a}")
print(f"10 in a: {10 in a}")
````

## Dictionaries

A **dictionary** stores key-value pairs. It's like a phone book — you look up a name (key) to get a number (value).

```python
# Dicts use curly braces with key: value pairs
person = {
    "name": "Ada",
    "age": 36,
    "city": "London"
}

# Access by key
person["name"]  # "Ada"

# Add or update
person["job"] = "Mathematician"
```

**ML relevance:** Model hyperparameters, config files, and JSON data from APIs are all dictionaries.

````python
# Dictionaries
person = {
    "name": "Ada",
    "age": 36,
    "city": "London"
}

print(f"person: {person}")
print(f"name: {person['name']}")
print(f"age: {person['age']}")

# Add / update
person["job"] = "Mathematician"
person["age"] = 37
print(f"after update: {person}")
````

````python
# Dict methods
d = {"a": 1, "b": 2, "c": 3}

print(f"keys: {list(d.keys())}")
print(f"values: {list(d.values())}")
print(f"items: {list(d.items())}")
print(f"get('a'): {d.get('a')}")
print(f"get('z', default): {d.get('z', 'not found')}")

# Iterate
for key, value in d.items():
    print(f"{key} -> {value}")
````

## List Comprehensions

A **list comprehension** is a concise way to build a list from an existing iterable. It replaces common `for` loop patterns.

```python
# Instead of:
squares = []
for x in range(10):
    squares.append(x**2)

# Write:
squares = [x**2 for x in range(10)]

# With a filter:
evens = [x for x in range(20) if x % 2 == 0]
```

**When to use:** When you're building a list by transforming/filtering another iterable. If the comprehension gets longer than one line, use a regular loop instead.

````python
# List comprehension
squares = [x**2 for x in range(10)]
print(f"squares: {squares}")

# With condition
evens = [x for x in range(20) if x % 2 == 0]
print(f"evens: {evens}")

# Equivalent to a for loop
squares_loop = []
for x in range(10):
    squares_loop.append(x**2)
print(f"squares_loop: {squares_loop}")
print(f"Same result: {squares == squares_loop}")
````

## Choosing the Right Structure

| Task | Best Structure | Why | Example |
|------|---------------|-----|---------|
| Ordered sequence, mutable | **List** | Can add/remove/sort | `[1, 2, 3]` |
| Fixed data | **Tuple** | Immutable, safe | `(3, 4)` coordinates |
| Unique items / membership | **Set** | O(1) lookup, dedup | `{'cat', 'dog'}` |
| Key-value lookup | **Dict** | O(1) by key | `{"name": "Ada"}` |
| Need both key and value iteration | **Dict** | `.items()` gives pairs | `for k, v in d.items()` |
| Need to preserve order + dedup | **List + Set** | Use set for tracking | `seen = set()` pattern |

## Experiment: Word Frequency

Count how often each word appears in a sentence using a dictionary. This is the foundation of **Bag of Words** models in NLP.

````python
# Word frequency counter
sentence = "the cat sat on the mat the cat ran"
words = sentence.split()

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

print(f"Words: {words}")
print(f"Frequency: {freq}")

# Most common word
most_common = max(freq, key=freq.get)
print(f"Most common: '{most_common}' ({freq[most_common]} times)")
````

## Common Mistakes

| Mistake | What goes wrong | How to fix it |
|---------|----------------|---------------|
| Using a list when a set is better | Slow O(n) membership checks | Use `set()` for dedup and membership |
| Mutable default argument | `def f(x, lst=[])` — shared across calls | Use `lst=None` then `if lst is None: lst = []` |
| Modifying a list while iterating | Skips items or errors | Iterate over a copy: `for item in lst[:]:` |
| Using mutable type as dict key | `TypeError: unhashable type: 'list'` | Use tuples as keys |
| Forgetting `.get()` for dicts | `KeyError` if key missing | Use `d.get(key, default)` |
| Slicing returns a new list | `b = a` is NOT a copy | Use `b = a.copy()` or `b = a[:]` |

````python
# BUG: mutable default argument
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("a"))  # ['a']
print(add_item("b"))  # ['a', 'b']  <- BUG! shared default

# FIX: use None as default
def add_item_fixed(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item_fixed("a"))  # ['a']
print(add_item_fixed("b"))  # ['b']  <- correct, fresh list each time
````

````python
# Demonstrate KeyError and fix
d = {"a": 1}
try:
    print(d["b"])
except KeyError as e:
    print(f"KeyError: {e}")
    print(f"Fixed with .get(): {d.get('b', 'default')}")
````

## Real-World Considerations

- **Performance**: Sets and dicts give O(1) lookup; lists are O(n). For large data, choose wisely.
- **Immutability**: Use tuples for data that must not change — it prevents bugs.
- **Readability**: List comprehensions are concise but don't over-nest them.
- **`collections` module**: `defaultdict`, `Counter`, `OrderedDict` are powerful tools built on these structures.

Write a function that takes a list of numbers and returns a new list with duplicates removed **while preserving the original order**.

````python
# Challenge: dedup preserving order
nums = [3, 1, 2, 3, 4, 1, 5, 2]

seen = set()
result = []
for n in nums:
    if n not in seen:
        seen.add(n)
        result.append(n)

print(f"original: {nums}")
print(f"deduped (order preserved): {result}")
````
