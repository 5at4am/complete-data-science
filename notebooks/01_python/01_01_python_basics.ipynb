## Variables and Data Types

A **variable** is a name that refers to a value. Python is **dynamically typed** — you don't declare the type; Python infers it.

```python
# Python figures out the type automatically:
x = 42          # int — Python sees a whole number
y = 3.14        # float — Python sees a decimal
name = "Ada"    # str — Python sees quotes
```

Core data types:

| Type | What it holds | Example | When to use |
|------|---------------|---------|-------------|
| `int` | Whole numbers | `42`, `-7`, `0` | Counts, indices, ages |
| `float` | Decimals | `3.14`, `-0.5` | Measurements, probabilities |
| `str` | Text | `"hello"`, `'world'` | Names, labels, file paths |
| `bool` | True/False | `True`, `False` | Conditions, flags |
| `None` | Nothing | `None` | Default values, missing data |

````python
# Variables and types
age = 25                # int
price = 19.99           # float
name = "Ada"            # str
is_student = True       # bool
nothing = None          # NoneType

print(f"age: {age}, type: {type(age)}")
print(f"price: {price}, type: {type(price)}")
print(f"name: {name}, type: {type(name)}")
print(f"is_student: {is_student}, type: {type(is_student)}")
print(f"nothing: {nothing}, type: {type(nothing)}")
````

### Type Conversion

Sometimes you need to convert between types. This is one of the most common sources of errors for beginners.

```python
# These are safe conversions:
int(42)        # int → int (no-op)
float(42)      # int → float: 42.0
str(42)        # int → str: "42"

# These can fail:
int("hello")   # ValueError: invalid literal
int("3.14")    # ValueError: can't convert float to int directly
int(float("3.14"))  # Correct: float first, then int → 3
```

````python
# Type conversion
x = "42"          # string
x_int = int(x)    # convert to int
x_float = float(x)  # convert to float

print(f"x: {x} (str), x_int: {x_int} (int), x_float: {x_float} (float)")

# Convert number to string
y = 3.14
y_str = str(y)
print(f"y: {y} (float), y_str: {y_str} (str)")
````

## Operators

### Arithmetic Operators

| Operator | Meaning | Example | ML use case |
|----------|---------|---------|-------------|
| `+` | addition | `5 + 3 = 8` | Combine feature values |
| `-` | subtraction | `5 - 3 = 2` | Compute errors (predicted - actual) |
| `*` | multiplication | `5 * 3 = 15` | Weighted sums |
| `/` | division | `5 / 2 = 2.5` | Normalize values |
| `//` | floor division | `5 // 2 = 2` | Integer bucketing |
| `%` | modulo (remainder) | `5 % 2 = 1` | Check divisibility, alternating logic |
| `**` | exponent | `2 ** 3 = 8` | Polynomial features |

**Key insight:** `/` always returns a float in Python 3. Use `//` when you need an integer result.

````python
# Arithmetic
a, b = 10, 3
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")
````

### Comparison and Logical Operators

Comparison operators return `True` or `False`. They're the building blocks of conditionals.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `5 > 3` | `True` |
| `<` | less than | `5 < 3` | `False` |
| `>=` | greater or equal | `5 >= 5` | `True` |
| `<=` | less or equal | `5 <= 3` | `False` |

**Logical operators** combine conditions:

| Operator | Meaning | Example | When to use |
|----------|---------|---------|-------------|
| `and` | both true | `(x > 0) and (x < 10)` | Range checks |
| `or` | at least one true | `(x < 0) or (x > 100)` | Boundary conditions |
| `not` | negate | `not is_valid` | Flip a flag |

````python
# Comparison and logical operators
x, y = 5, 10

print(f"x == y: {x == y}")
print(f"x != y: {x != y}")
print(f"x < y: {x < y}")
print(f"x > y: {x > y}")

print(f"(x < y) and (x > 0): {(x < y) and (x > 0)}")
print(f"(x > y) or (x > 0): {(x > y) or (x > 0)}")
print(f"not (x > y): {not (x > y)}")
````

## Control Flow: Conditionals

`if`/`elif`/`else` lets you make decisions. The indentation is **mandatory** — it defines the code block.

```python
if condition:
    # runs if condition is True
elif other_condition:
    # runs if first condition was False AND this one is True
else:
    # runs if all conditions were False
```

**ML relevance:** Decision trees literally implement this pattern — each node is an if/else split on a feature.

````python
# Conditionals
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
````

## Control Flow: Loops

### `for` loops — iterate over a sequence

Use `for` when you know **how many times** to iterate (or have a collection to walk through).

```python
for item in collection:
    # do something with item
```

**Common patterns:**
- `range(n)` — iterate 0 to n-1
- `range(start, stop, step)` — custom range
- `for item in list` — iterate over elements directly

````python
# for loop over a range
print("Counting 0 to 4:")
for i in range(5):
    print(i, end=" ")
print()

# for loop over a list
print("\nIterating over a list:")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

````python
# Sum of numbers 1 to 100
total = 0
for i in range(1, 101):
    total += i
print(f"Sum of 1 to 100: {total}")

# Verify with formula n(n+1)/2
formula = 100 * 101 // 2
print(f"Formula check: {formula}")
print(f"Match: {total == formula}")
````

### `while` loops — repeat until a condition is false

Use `while` when you **don't know** how many iterations you need — you loop until something changes.

```python
while condition:
    # do something
    # IMPORTANT: something must eventually make condition False!
```

**Warning:** If the condition never becomes `False`, you get an **infinite loop** — the most common beginner bug. Always ensure your loop variable gets updated.

````python
# while loop
count = 0
while count < 5:
    print(f"count: {count}")
    count += 1

print(f"Loop ended, count = {count}")
````

### `break` and `continue`

| Keyword | What it does | Analogy |
|---------|-------------|---------|
| `break` | Exit the loop entirely | Pull the fire alarm — leave the building |
| `continue` | Skip to the next iteration | Skip a song in your playlist — move to the next one |

**When to use:**
- `break`: You found what you're looking for (early exit)
- `continue`: One item is invalid but the rest are fine (skip bad data)

````python
# break: stop at 3
print("break example:")
for i in range(10):
    if i == 3:
        break
    print(i, end=" ")
print()

# continue: skip even numbers
print("continue example (skip evens):")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
````

## Experiment: FizzBuzz

Let's combine everything into a classic programming challenge: **FizzBuzz**.

For numbers 1 to 20:
- If divisible by 3, print "Fizz"
- If divisible by 5, print "Buzz"
- If divisible by both, print "FizzBuzz"
- Otherwise, print the number

This is a great exercise because it tests: conditionals, modulo operator, loops, and string formatting — all in one problem.

````python
# FizzBuzz
for i in range(1, 21):
    if i % 15 == 0:
        print("FizzBuzz", end=" ")
    elif i % 3 == 0:
        print("Fizz", end=" ")
    elif i % 5 == 0:
        print("Buzz", end=" ")
    else:
        print(i, end=" ")
print()
````

## Common Mistakes

| Mistake | What goes wrong | How to fix it |
|---------|----------------|---------------|
| Forgetting to update loop variable | Infinite loop — program hangs | Always increment/decrement the variable |
| Mixing types in operations | `TypeError: can only concatenate str to str` | Convert types explicitly: `int("5") + 3` |
| Inconsistent indentation | `IndentationError` — code won't run | Use 4 spaces everywhere (Python enforces this) |
| Using `=` instead of `==` | Assignment instead of comparison | `=` assigns, `==` compares |
| Off-by-one in `range()` | `range(1, 5)` gives 1,2,3,4 — not 5 | Remember: stop is exclusive |
| `float` comparison | `0.1 + 0.2 == 0.3` is `False` | Use `abs(a - b) < 1e-9` instead |

### Debugging: Common Errors

### NameError — variable not defined
```python
print(undefined_var)  # NameError: name 'undefined_var' is not defined
```
**Fix:** Make sure the variable is defined before you use it.

### TypeError — wrong type operation
```python
"5" + 3  # TypeError: can only concatenate str (not "int") to str
```
**Fix:** Convert types: `int("5") + 3` → `8`

### IndentationError — inconsistent indentation
```python
if True:
print("bad")  # IndentationError
```
**Fix:** Use exactly 4 spaces for each indentation level. Python uses indentation as syntax — it's not optional.

````python
# BUG: infinite loop (commented out to avoid hanging)
# count = 0
# while count < 5:
#     print(count)   # forgot count += 1 -> infinite loop!

# Correct version:
count = 0
while count < 5:
    print(count, end=" ")
    count += 1  # this line prevents the infinite loop
print()
print("Correct: loop terminated because count is updated.")
````

````python
# Demonstrate a TypeError and fix it
try:
    result = "5" + 3
except TypeError as e:
    print(f"TypeError: {e}")
    print("Fix: convert the string to int first")
    result = int("5") + 3
    print(f"Fixed result: {result}")
````

## Real-World Considerations

- **Readability matters**: use meaningful variable names (`total_score` not `ts`). Your future self will thank you.
- **PEP 8**: Python's style guide — 4-space indentation, snake_case names, spaces around operators.
- **Comments**: explain *why*, not *what*. The code explains what; your comment explains why.
- **f-strings**: use them for readable string formatting: `f"{x}"` instead of `"%d" % x` or `"{}".format(x)`.

### Decision Guide: `for` vs `while`

| Scenario | Use | Why |
|----------|-----|-----|
| Iterate over a list | `for item in list` | Direct, readable |
| Count from 0 to N | `for i in range(N)` | Handles the counter |
| Repeat until condition met | `while condition` | Unknown iteration count |
| Read input until 'quit' | `while True` + `break` | Interactive loop |
