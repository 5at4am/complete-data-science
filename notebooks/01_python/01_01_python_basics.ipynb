## 6. Variables and Data Types

A **variable** is a name that refers to a value. Python is **dynamically typed** — you don't declare the type; Python infers it.

Core data types:

- `int` — integers (e.g., `42`)
- `float` — decimals (e.g., `3.14`)
- `str` — text (e.g., `"hello"`)
- `bool` — True/False
- `None` — absence of value

````python
# Variables and types
age = 25                # int
price = 19.99           # float
name = "Ada"            # str
is_student = True       # bool
nothing = None          # NoneType

print(f"age: {age}, type: {type(age)}")
print(f"price: {price}, type: {type(price)}")
print(f"name: {name}, type: {type(name)}")
print(f"is_student: {is_student}, type: {type(is_student)}")
print(f"nothing: {nothing}, type: {type(nothing)}")
````

### Type Conversion

Sometimes you need to convert between types.

````python
# Type conversion
x = "42"          # string
x_int = int(x)    # convert to int
x_float = float(x)  # convert to float

print(f"x: {x} (str), x_int: {x_int} (int), x_float: {x_float} (float)")

# Convert number to string
y = 3.14
y_str = str(y)
print(f"y: {y} (float), y_str: {y_str} (str)")
````

## 7. Operators

### Arithmetic Operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `+` | addition | `5 + 3 = 8` |
| `-` | subtraction | `5 - 3 = 2` |
| `*` | multiplication | `5 * 3 = 15` |
| `/` | division | `5 / 2 = 2.5` |
| `//` | floor division | `5 // 2 = 2` |
| `%` | modulo (remainder) | `5 % 2 = 1` |
| `**` | exponent | `2 ** 3 = 8` |

````python
# Arithmetic
a, b = 10, 3
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")
````

### Comparison Operators

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal |
| `<=` | less than or equal |

### Logical Operators

| Operator | Meaning |
|----------|---------|
| `and` | both true |
| `or` | at least one true |
| `not` | negate |

````python
# Comparison and logical operators
x, y = 5, 10

print(f"x == y: {x == y}")
print(f"x != y: {x != y}")
print(f"x < y: {x < y}")
print(f"x > y: {x > y}")

print(f"(x < y) and (x > 0): {(x < y) and (x > 0)}")
print(f"(x > y) or (x > 0): {(x > y) or (x > 0)}")
print(f"not (x > y): {not (x > y)}")
````

## 8. Control Flow: Conditionals

`if`/`elif`/`else` lets you make decisions.

````python
# Conditionals
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
````

## 9. Control Flow: Loops

### `for` loops — iterate over a sequence

````python
# for loop over a range
print("Counting 0 to 4:")
for i in range(5):
    print(i, end=" ")
print()

# for loop over a list
print("\nIterating over a list:")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

````python
# Sum of numbers 1 to 100
total = 0
for i in range(1, 101):
    total += i
print(f"Sum of 1 to 100: {total}")

# Verify with formula n(n+1)/2
formula = 100 * 101 // 2
print(f"Formula check: {formula}")
print(f"Match: {total == formula}")
````

### `while` loops — repeat until a condition is false

````python
# while loop
count = 0
while count < 5:
    print(f"count: {count}")
    count += 1

print(f"Loop ended, count = {count}")
````

### `break` and `continue`

- `break` — exit the loop entirely
- `continue` — skip to the next iteration

````python
# break: stop at 3
print("break example:")
for i in range(10):
    if i == 3:
        break
    print(i, end=" ")
print()

# continue: skip even numbers
print("continue example (skip evens):")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
````

## 10. Experiment: FizzBuzz

Let's combine everything into a classic programming challenge: **FizzBuzz**.

For numbers 1 to 20:
- If divisible by 3, print "Fizz"
- If divisible by 5, print "Buzz"
- If divisible by both, print "FizzBuzz"
- Otherwise, print the number

````python
# FizzBuzz
for i in range(1, 21):
    if i % 15 == 0:
        print("FizzBuzz", end=" ")
    elif i % 3 == 0:
        print("Fizz", end=" ")
    elif i % 5 == 0:
        print("Buzz", end=" ")
    else:
        print(i, end=" ")
print()
````

## 11. Failure Case: Infinite Loop

A common bug is an **infinite loop** — the condition never becomes false. This happens when you forget to update the loop variable.

````python
# BUG: infinite loop (commented out to avoid hanging)
# count = 0
# while count < 5:
#     print(count)   # forgot count += 1 -> infinite loop!

# Correct version:
count = 0
while count < 5:
    print(count, end=" ")
    count += 1  # this line prevents the infinite loop
print()
print("Correct: loop terminated because count is updated.")
````

## 12. Debugging: Common Errors

### NameError — variable not defined
```python
print(undefined_var)  # NameError
```

### TypeError — wrong type operation
```python
"5" + 3  # TypeError: can only concatenate str
```

### IndentationError — inconsistent indentation
```python
if True:
print("bad")  # IndentationError
```

Python uses **indentation** (4 spaces) to define blocks. This is not optional — it's part of the syntax.

````python
# Demonstrate a TypeError and fix it
try:
    result = "5" + 3
except TypeError as e:
    print(f"TypeError: {e}")
    print("Fix: convert the string to int first")
    result = int("5") + 3
    print(f"Fixed result: {result}")
````

## 13. Real-World Considerations

- **Readability matters**: use meaningful variable names (`total_score` not `ts`).
- **PEP 8**: Python's style guide — 4-space indentation, snake_case names.
- **Comments**: explain *why*, not *what*.
- **f-strings**: use them for readable string formatting (`f"{x}"`).

## 14. Common Mistakes

- Forgetting to update loop variables (infinite loops)
- Mixing types (TypeError)
- Inconsistent indentation
- Using `=` instead of `==` in conditions
- Off-by-one errors in `range()`

## 15. When NOT to Use

- Don't use `while` when `for` is clearer (prefer `for` for known iteration counts).
- Don't write long, complex conditionals — break them into functions.
