## 5. Mental Model

LCEL = pipe-based composition.
chain = prompt | llm | parser
Supports: .invoke(), .stream(), .batch(), .ainvoke()

````python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
print("Libraries loaded.")
````

## 6. Basic LCEL Chain

````python
llm = ChatGroq(model="qwen/qwen3.8-27b", temperature=0)

prompt = ChatPromptTemplate.from_template("Explain {topic} in {style} style.")

chain = prompt | llm | StrOutputParser()

result = chain.invoke({"topic": "RAG", "style": "concise"})
print("Result:", result[:300])
````

## 7. RunnablePassthrough

````python
chain = (
    {"topic": RunnablePassthrough(), "style": lambda x: "concise"} 
    | prompt 
    | llm 
    | StrOutputParser()
)

result = chain.invoke("transformers")
print("Result:", result[:300])
````

## 8. Streaming

````python
print("Streaming response:")
for chunk in chain.stream("langchain"):
    print(chunk, end="", flush=True)
print()
````

## 9. Reusable Components

````python
def create_analysis_chain(topic_prompt):
    prompt = ChatPromptTemplate.from_template(topic_prompt)
    return prompt | llm | StrOutputParser()

summary_chain = create_analysis_chain("Summarize: {text}")
qa_chain = create_analysis_chain("Answer: {text}")

print("Summary:", summary_chain.invoke("LangChain is a framework for LLM apps.")[:200])
print("QA:", qa_chain.invoke("What is LangChain?")[:200])
````
