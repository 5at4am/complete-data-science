## 6. The Document Object

````python
from langchain_core.documents import Document

doc = Document(
    page_content="LangChain is a framework for building LLM applications.",
    metadata={"source": "example.txt", "page": 1}
)

print("Content:", doc.page_content)
print("Metadata:", doc.metadata)
print("Type:", type(doc))
````

## 7. Loading Text Files

````python
sample_text = "LangChain is a framework for building applications powered by language models.\nIt provides tools for prompt management, chaining, memory, and more.\nLangChain supports multiple LLM providers including OpenAI, Anthropic, and Groq.\nThe framework is designed for production use with built-in observability.\n"

sample_path = r"D:\CODE\complete ml\temp_sample.txt"
with open(sample_path, "w") as f:
    f.write(sample_text)

# Manual loading (works without langchain_community)
with open(sample_path, "r") as f:
    content = f.read()

doc = Document(page_content=content, metadata={"source": sample_path})
print("Loaded document from", sample_path)
print("Content preview:", doc.page_content[:100])
````

## 8. Loading CSV Files

````python
import csv

csv_data = "name,role,department\nAlice,Engineer,Engineering\nBob,Designer,Design\nCharlie,Manager,Operations\n"

csv_path = r"D:\CODE\complete ml\temp_sample.csv"
with open(csv_path, "w") as f:
    f.write(csv_data)

# Load CSV as Documents
docs = []
with open(csv_path, "r") as f:
    reader = csv.DictReader(f)
    for i, row in enumerate(reader):
        content = ", ".join(f"{k}: {v}" for k, v in row.items())
        docs.append(Document(page_content=content, metadata={"source": csv_path, "row": i}))

print("Loaded " + str(len(docs)) + " documents from CSV")
for doc in docs:
    print("  -", doc.page_content)
````

## 9. Creating Documents from Strings

````python
docs = [
    Document(page_content="First paragraph about AI.", metadata={"index": 0, "topic": "AI"}),
    Document(page_content="Second paragraph about ML.", metadata={"index": 1, "topic": "AI"}),
    Document(page_content="Third paragraph about DL.", metadata={"index": 2, "topic": "AI"})
]

print("Created " + str(len(docs)) + " documents")
for d in docs:
    print("  -", d.metadata, ":", d.page_content[:40])
````

## 10. Batch Loading

````python
import glob

for i in range(3):
    path = r"D:\CODE\complete ml\temp_doc_" + str(i) + ".txt"
    with open(path, "w") as f:
        f.write("Document " + str(i) + " content about topic " + str(i) + ".")

txt_files = glob.glob(r"D:\CODE\complete ml\temp_doc_*.txt")
all_docs = []
for f in txt_files:
    with open(f, "r") as fh:
        content = fh.read()
    all_docs.append(Document(page_content=content, metadata={"source": f}))

print("Loaded " + str(len(all_docs)) + " total documents from " + str(len(txt_files)) + " files")
````
