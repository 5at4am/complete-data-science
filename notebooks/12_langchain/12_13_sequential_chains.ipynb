## 1. What Are We Solving?

Complex workflows require chaining multiple LLM calls. Sequential chains pass output from one step to the next.

## 2. Why Does This Matter?

Many tasks need multiple steps: summarize then translate, analyze then advise.

## 4. Learning Objectives

- Pipe outputs between steps
- Use RunnableParallel for concurrent steps

## 5. Mental Model

Sequential chain: step1 | step2 | step3
Each step transforms the output for the next.

````python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
print("Libraries loaded.")
````

## 6. Basic Sequential Chain

````python
llm = ChatGroq(model="qwen/qwen3.8-27b", temperature=0)

analyze_prompt = ChatPromptTemplate.from_template(
    "Analyze this topic in 2 sentences: {topic}"
)

example_prompt = ChatPromptTemplate.from_template(
    "Given this analysis:\n{analysis}\n\nGive 3 practical examples." 
)

chain = analyze_prompt | llm | StrOutputParser() | example_prompt | llm | StrOutputParser()

result = chain.invoke({"topic": "machine learning"})
print("Result:", result[:300])
````

## 7. RunnableParallel

````python
from langchain_core.runnables import RunnableParallel

summary_prompt = ChatPromptTemplate.from_template("Summarize in 1 sentence: {text}")
keywords_prompt = ChatPromptTemplate.from_template("Extract 3 keywords from: {text}")
combine_prompt = ChatPromptTemplate.from_template(
    "Summary: {summary}\nKeywords: {keywords}\n\nCombine these into a brief overview."
)

parallel = RunnableParallel(
    summary=summary_prompt | llm | StrOutputParser(),
    keywords=keywords_prompt | llm | StrOutputParser()
)

chain = parallel | combine_prompt | llm | StrOutputParser()
result = chain.invoke({"text": "LangChain simplifies building LLM applications with chains, agents, and retrieval."})
print("Combined:", result[:300])
````

## 8. Common Mistakes

1. Passing wrong variable names between steps
2. Not handling errors in chains
3. Chains too long (compounding errors)
4. Not using parallel when possible
