## 5. Mental Model

LangServe = LangChain + FastAPI.
One line to deploy any chain as an API.

````python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
print("Libraries loaded.")
````

## 6. Build a Deployable Chain

````python
llm = ChatGroq(model="qwen/qwen3.8-27b", temperature=0)

prompt = ChatPromptTemplate.from_template(
    "You are a helpful assistant. Answer the question:\n\nQuestion: {question}"
)

chain = prompt | llm | StrOutputParser()

result = chain.invoke({"question": "What is LangChain?"})
print("Chain works:", result[:100])
````

## 9. Input/Output Validation

````python
from pydantic import BaseModel

class QuestionInput(BaseModel):
    question: str

class AnswerOutput(BaseModel):
    answer: str
    source: str = "langchain"

print("Input schema:", QuestionInput.model_json_schema())
print("Output schema:", AnswerOutput.model_json_schema())
````
