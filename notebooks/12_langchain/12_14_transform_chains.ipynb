## 1. What Are We Solving?

Not every step needs an LLM. Transform chains apply Python functions to data between LLM calls.

## 2. Why Does This Matter?

Data transformation, formatting, and validation are essential in production pipelines.

## 5. Mental Model

Transform = Python function in a chain.
input -> transform -> LLM -> transform -> output.

````python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
print("Libraries loaded.")
````

## 6. Basic Transform

````python
llm = ChatGroq(model="qwen/qwen3.8-27b", temperature=0)

def to_uppercase(input_dict):
    return {"text": input_dict["text"].upper()}

chain = RunnableLambda(to_uppercase) | ChatPromptTemplate.from_template("Explain this: {text}") | llm | StrOutputParser()

result = chain.invoke({"text": "machine learning"})
print("Result:", result[:200])
````

## 7. Data Processing Pipeline

````python
def extract_key_points(input_dict):
    text = input_dict["text"]
    lines = text.strip().split("\n")
    key_points = [l.strip() for l in lines if l.strip()]
    numbered = "\n".join(str(i+1) + ". " + p for i, p in enumerate(key_points))
    return {"points": numbered}

def format_prompt(input_dict):
    return {"text": "Organize these key points into a structured summary:\n" + input_dict["points"]}

chain = RunnableLambda(extract_key_points) | RunnableLambda(format_prompt) | ChatPromptTemplate.from_template("{text}") | llm | StrOutputParser()

result = chain.invoke({"text": "AI is transforming healthcare\nML improves diagnosis\nDL enables image analysis"})
print("Result:", result[:300])
````

## 8. Error Handling

````python
def safe_transform(input_dict):
    try:
        result = input_dict["text"].split(".")[0]
        return {"summary": result}
    except Exception as e:
        return {"summary": "Error processing text: " + str(e)}

chain = RunnableLambda(safe_transform) | ChatPromptTemplate.from_template("Expand this: {summary}") | llm | StrOutputParser()

result = chain.invoke({"text": "AI is great. ML is powerful. DL is amazing."})
print("Result:", result[:200])
````

## 9. Common Mistakes

1. Not handling exceptions in transforms
2. Returning wrong key names
3. Mutating input dictionaries
4. Heavy computation in transforms
