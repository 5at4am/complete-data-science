````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine, load_digits
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
import warnings
warnings.filterwarnings("ignore")
np.random.seed(42)
print("Libraries loaded.")
````

## 6. MLP on Wine Dataset

````python
wine = load_wine()
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

mlp = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=42)
mlp.fit(X_train_s, y_train)

print("Train accuracy:", round(mlp.score(X_train_s, y_train), 4))
print("Test accuracy:", round(mlp.score(X_test_s, y_test), 4))
````

## 7. Architecture Search

````python
archs = [(16,), (32, 16), (64, 32), (128, 64, 32), (256, 128, 64, 32)]
results = []
for arch in archs:
    mlp = MLPClassifier(hidden_layer_sizes=arch, max_iter=500, random_state=42)
    scores = cross_val_score(mlp, X_train_s, y_train, cv=5)
    results.append({"Architecture": str(arch), "CV Accuracy": round(scores.mean(), 4)})

print(pd.DataFrame(results).to_string(index=False))
````

## 8. Learning Curves

````python
mlp = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=42)
mlp.fit(X_train_s, y_train)

plt.figure(figsize=(10, 5))
plt.plot(mlp.loss_curve_, "b-", label="Training loss")
# NOTE: validation_scores_ exists as an attribute but is None unless early_stopping
# is enabled with a validation_fraction. Check the VALUE, not just the attribute.
if mlp.validation_scores_ is not None:
    plt.plot(mlp.validation_scores_, "r-", label="Validation score")
plt.xlabel("Iteration")
plt.ylabel("Loss / Score")
plt.title("MLP Learning Curve")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("mlp_learning_curve.png", dpi=100, bbox_inches="tight")
plt.show()
````

## 9. Compare with Traditional Models

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "MLP (64,32)": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=42)
}

for name, model in models.items():
    scores = cross_val_score(model, X_train_s, y_train, cv=5)
    print(name + ": " + str(round(scores.mean(), 4)) + " (+/- " + str(round(scores.std(), 4)) + ")")
````
