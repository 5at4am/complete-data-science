## 1. What Are We Solving?

Clustering is **unsupervised learning**: we group data points into clusters without labels. We cover K-Means, Hierarchical, and DBSCAN.

## 2. Why Does This Matter?

Clustering is used for customer segmentation, anomaly detection, image compression, and exploratory analysis. Understanding the tradeoffs between algorithms is key.

## 5. Mental Model

- **K-Means**: partition into k clusters by minimizing within-cluster variance. Needs k.
- **Hierarchical**: build a tree of clusters (agglomerative).
- **DBSCAN**: density-based, finds arbitrary shapes and outliers. Needs eps and min_samples.


## 6. Generate Data

Create blobs and a dataset with noise.


````python
import matplotlib
matplotlib.use('Agg')
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.preprocessing import StandardScaler

np.random.seed(42)
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)
print(f"Blobs: {X.shape}")

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], alpha=0.5)
plt.title("Blobs (true clusters)")
plt.show()
````

## 7. K-Means

K-Means partitions data into k clusters.


````python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', alpha=0.5)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='x', s=200, label="Centroids")
plt.title("K-Means (k=3)")
plt.legend()
plt.show()
print(f"K-Means inertia: {kmeans.inertia_:.3f}")
````

## 8. Choosing k: Elbow Method

Plot inertia vs k to find the 'elbow'.


````python
inertias = []
for k in range(1, 8):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    inertias.append(km.inertia_)

plt.figure(figsize=(6, 4))
plt.plot(range(1, 8), inertias, marker='o')
plt.xlabel("k")
plt.ylabel("Inertia")
plt.title("Elbow method")
plt.show()
print("The elbow at k=3 suggests 3 clusters.")
````

## 9. Hierarchical Clustering

Agglomerative clustering builds a hierarchy.


````python
hier = AgglomerativeClustering(n_clusters=3)
hier_labels = hier.fit_predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=hier_labels, cmap='viridis', alpha=0.5)
plt.title("Hierarchical Clustering")
plt.show()
print("Hierarchical clustering groups data into a tree.")
````

## 10. DBSCAN

DBSCAN finds clusters of arbitrary shape and identifies outliers.


````python
# Moons data - K-Means fails, DBSCAN works
Xm, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
Xm = StandardScaler().fit_transform(Xm)

km = KMeans(n_clusters=2, random_state=42, n_init=10).fit_predict(Xm)
db = DBSCAN(eps=0.3, min_samples=5).fit_predict(Xm)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(Xm[:, 0], Xm[:, 1], c=km, cmap='viridis', alpha=0.5)
plt.title("K-Means on moons (fails)")
plt.subplot(1, 2, 2)
plt.scatter(Xm[:, 0], Xm[:, 1], c=db, cmap='viridis', alpha=0.5)
plt.title("DBSCAN on moons (works)")
plt.show()
print(f"DBSCAN found {len(set(db)) - (1 if -1 in db else 0)} clusters")
print(f"Outliers (label -1): {(db == -1).sum()}")
````

## 11. Failure Case: K-Means on Non-Spherical Data

K-Means assumes spherical clusters of similar size.


````python
print("K-Means fails on the moons data because clusters are not spherical.")
print("DBSCAN handles arbitrary shapes and finds outliers.")
print("\nChoose the algorithm based on data shape:")
print("- Spherical, known k: K-Means")
print("- Hierarchy needed: Hierarchical")
print("- Arbitrary shapes/outliers: DBSCAN")
````

## 12. Debugging: Common Errors

- **Wrong k**: use elbow/silhouette.
- **Not scaling**: distance-based, scale matters.
- **K-Means on non-spherical data**: wrong algorithm.

## 13. Real-World Considerations

- Scale features before clustering.
- Use silhouette score to evaluate.
- Clustering is exploratory - validate with domain knowledge.

## 14. Common Mistakes

- Not scaling.
- Assuming K-Means works for all shapes.

## 15. When NOT to Use

- When you need labeled predictions.
- When clusters are highly non-spherical (use DBSCAN).

## 16. Challenge

Compute the silhouette score for K-Means with different k values.


````python
# Challenge: silhouette score
from sklearn.metrics import silhouette_score
for k in [2, 3, 4, 5]:
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    sil = silhouette_score(X, km.labels_)
    print(f"k={k}: silhouette={sil:.3f}")
print("\nHigher silhouette = better-defined clusters.")
````
