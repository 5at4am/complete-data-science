## 1. What Are We Solving?

Git is a **version control system** that tracks changes to files over time. It lets you:

- Save snapshots of your work (commits)
- Revert to previous versions
- Work on branches without breaking things
- Collaborate with others
- **Recover from mistakes** (critical for this curriculum)

**The problem:** Without version control, your only backup is copy-pasting files (`model_v2_final_FINAL.py`). One accidental deletion, one bad edit, one corrupted file — and hours of work are gone. Git gives you a time machine: every saved state is recoverable forever.

---

## 2. Why Does It Matter?

| Scenario | Without Git | With Git |
|----------|-------------|----------|
| You delete a file by accident | File is gone forever | `git checkout -- <file>` restores it |
| You break your code and can't fix it | Start over from scratch | `git revert` undoes the bad commit |
| You want to try a risky experiment | Copy entire project folder | Create a branch, experiment freely |
| A colleague asks what you changed | "Uh, I think I did..." | `git log` shows every change with context |
| Your laptop dies | Everything is lost | Push to GitHub = backup in the cloud |

**Real-world example:** A student spent 3 hours tuning a model. They accidentally ran a cell that overwrote their best parameters. Without Git, they had to redo everything. With Git, `git checkout` restored the file in 2 seconds.

This learning system is designed so that **losing files does not destroy progress**. Git is the backbone of that recovery system. Every completed unit is committed, so you can always recover.

## 3. Mental Model

```text
Your Computer                          GitHub (Cloud)
┌──────────────────┐                  ┌──────────────────┐
│                  │   git push       │                  │
│  Working Dir ────┼─────────────────→│  Remote Repo     │
│     (modified)   │                  │                  │
│       │          │   git pull       │  Backup of all   │
│       ▼          │←─────────────────│  your commits    │
│  Staging Area    │                  │                  │
│     (staged)     │                  └──────────────────┘
│       │          │
│       ▼          │
│  Repository      │
│  (committed)     │
│                  │
│  ══════════════  │
│  TIME MACHINE    │ ← You can go back to any commit
│  ══════════════  │
└──────────────────┘
```

**Analogy:** Git is like a video game save system. You can save at any point (commit), load any previous save (checkout), create alternate timelines (branches), and merge timelines back together (merge).

## 6. Core Concepts

### The Three States

```
Working Directory → Staging Area → Repository
     (modified)        (staged)        (committed)
```

| State | What It Means | How to Get There |
|-------|--------------|------------------|
| **Working Directory** | Files as they are on disk | Edit files normally |
| **Staging Area** | Files marked for next commit | `git add <file>` |
| **Repository** | Permanent snapshot | `git commit -m "message"` |

Think of it as a funnel:
1. You edit files (working directory)
2. You choose which changes to save (staging area)
3. You save a snapshot (repository/commit)

### Key Commands

```bash
git init              # create a repo
git status            # check state
git add <file>        # stage a file
git add .             # stage ALL changes
git commit -m "msg"   # commit staged files
git log               # view history
git log --oneline     # compact history
git diff              # view unstaged changes
git diff --staged     # view staged changes
```

### What `git status` Tells You

```
On branch main
Changes to be committed:       ← staged (green)
  new file:   notebook.ipynb

Changes not staged for commit: ← modified but not staged (red)
  modified:   README.md

Untracked files:               ← new, never tracked (red)
  data.csv
```

This is your dashboard. Check it often.

## 8. Branching

Branches let you work on features, fixes, or experiments **without directly changing the `main` branch**.

Think of a branch as a separate workspace:

```text
main
 │
 ├── commit A
 ├── commit B
 │
 └── feature/login
       ├── commit C
       └── commit D
```

You can develop your feature independently and merge it back into `main` when it is ready.

### 8.1 Check Existing Branches

```bash
git branch
```

The `*` shows the branch you are currently on.

```text
* main
  feature/login
```

---

### 8.2 Create a New Branch

Modern Git syntax:

```bash
git switch -c feature/login
```

This creates the branch **and switches to it**.

Older syntax:

```bash
git checkout -b feature/login
```

> **Recommended:** Learn `git switch` for branch operations. `git checkout` is still important because you will encounter it in existing projects.

---

### 8.3 Make Changes and Commit

After working on the feature:

```bash
git status

git add .

git commit -m "Add login feature"
```

Now the commit belongs to your `feature/login` branch, not directly to `main`.

---

### 8.4 Switch Between Branches

Switch to `main`:

```bash
git switch main
```

Switch back to your feature:

```bash
git switch feature/login
```

You can check where you are with:

```bash
git branch
```

---

### 8.5 Merge a Branch

Once the feature is complete, switch to `main`:

```bash
git switch main
```

Make sure `main` is up to date:

```bash
git pull
```

Then merge the feature:

```bash
git merge feature/login
```

The result:

```text
Before:

main
 A---B

feature/login
      \
       C---D


After:

main
 A---B-------M
      \     /
       C---D
```

`M` represents the merge commit when Git needs one.

---

### 8.6 Delete the Feature Branch

After successfully merging:

```bash
git branch -d feature/login
```

This deletes the local branch.

If you also pushed the branch to GitHub:

```bash
git push origin --delete feature/login
```

---

### 8.7 Push a New Branch to GitHub

A newly created local branch does not automatically exist on GitHub.

Push it with:

```bash
git push -u origin feature/login
```

The `-u` sets the upstream branch, so later you can simply use:

```bash
git push
```

and

```bash
git pull
```

---

### 8.8 Typical Feature Workflow

A common workflow looks like this:

```bash
# Start from an updated main branch
git switch main
git pull

# Create feature branch
git switch -c feature/login

# Work on the feature
# ...

# Save your work
git add .
git commit -m "Add login feature"

# Push branch to GitHub
git push -u origin feature/login

# After the feature is reviewed/ready
git switch main
git pull

# Merge feature
git merge feature/login

# Push updated main
git push

# Delete local branch
git branch -d feature/login
```

---

### 8.9 What About Merge Conflicts?

Sometimes both `main` and your feature branch modify the same part of a file.

Git may report:

```text
CONFLICT (content): Merge conflict
```

Git will mark the conflicting section:

```text
<<<<<<< HEAD
code from main
=======
code from feature/login
>>>>>>> feature/login
```

You must manually decide which code to keep, then:

```bash
git add .
git commit
```

> **Important:** Do not blindly delete one side. Understand both changes before resolving a conflict.

---

### 8.10 Branch Naming

Use descriptive names instead of:

```text
branch1
test
new
abc
mybranch
```

Prefer:

```text
feature/login
feature/payment
feature/user-profile

fix/login-error
fix/navbar-overflow

docs/git-guide
refactor/auth-service
```

Common prefixes:

| Prefix | Purpose |
|--------|---------|
| `feature/` | New functionality |
| `fix/` | Bug fix |
| `docs/` | Documentation |
| `refactor/` | Code restructuring |
| `test/` | Testing-related work |
| `chore/` | Maintenance/configuration |

## 9. Recovering from Mistakes

Git's most powerful feature is **recovery**. Here are the common scenarios:

### Undo uncommitted changes

```bash
git checkout -- <file>   # discard working dir changes
git reset <file>         # unstage
```

### Revert a commit

```bash
git revert <commit>      # create a new commit undoing it
```

### Go back in history

```bash
git log --oneline        # find the commit
git checkout <commit>    # inspect (detached HEAD — be careful)
```

### Restore a deleted file

```bash
git checkout <commit> -- <file>
```

### See what changed

```bash
git diff                 # unstaged changes
git diff --staged        # staged changes
git diff main..feature   # diff between branches
```

| Situation | Command | What It Does |
|-----------|---------|--------------|
| Changed file, want to undo | `git checkout -- <file>` | Discards changes, restores last committed version |
| Staged wrong file | `git reset HEAD <file>` | Unstages without discarding changes |
| Bad commit already pushed | `git revert <hash>` | Creates new commit that undoes the bad one |
| Want to see history | `git log --oneline` | Shows compact commit history |
| Want to see what changed | `git diff` | Shows line-by-line differences |

## 10. .gitignore

`.gitignore` tells Git which files to ignore. This project ignores:

- `.venv/` (virtual environment)
- `.env` (secrets)
- `__pycache__/`
- `datasets/raw/*` (large data)
- model artifacts
- logs

**Never commit secrets** (API keys, tokens). They go in `.env` (gitignored).

### What to Gitignore

| File/Folder | Why |
|-------------|-----|
| `.venv/` | Environment-specific, huge, reproducible from pyproject.toml |
| `.env` | Contains secrets — API keys, tokens, passwords |
| `__pycache__/` | Auto-generated Python bytecode |
| `*.ipynb_checkpoints/` | Jupyter auto-save folders |
| `datasets/raw/` | Large data files — use DVC or download scripts instead |
| `model_artifacts/` | Trained models are large and reproducible |
| `*.log` | Log files are temporary |
| `.DS_Store` | macOS system file (not relevant to project) |

### .gitignore Format

```gitignore
# Comment line
.venv/
.env
__pycache__/
*.pyc
datasets/raw/
*.log
```

> **Important:** `.gitignore` only affects untracked files. If you already committed a file, adding it to `.gitignore` won't remove it from history. Use `git rm --cached <file>` to stop tracking it.

## 11. Commit Convention

This project uses meaningful commits:

```
phase-00-environment-unit-01-complete
phase-01-foundations-unit-03-complete
phase-03-ml-unit-02-complete
```

### Good vs Bad Commit Messages

| Bad | Good | Why |
|-----|------|-----|
| `fix` | `fix: resolve data loading error in unit 02` | Describes what was fixed |
| `update` | `phase-01-foundations-unit-03-complete` | Follows project convention |
| `wip` | `feat: add confusion matrix visualization` | Clear intent and scope |
| `asdf` | `docs: update README with setup instructions` | Meaningful and searchable |

### When to Commit

| Action | When |
|--------|------|
| Complete a unit | Commit immediately |
| Finish a major code change | Commit with descriptive message |
| Before risky experiment | Commit current state first |
| End of work session | Commit everything with clear message |

## 12. Troubleshooting Guide

| Symptom | Cause | Verify | Fix |
|---------|-------|--------|-----|
| `fatal: not a git repository` | You're not in a Git repo directory | `pwd` and `ls -la` to check for `.git/` | Run `git init` or navigate to repo |
| `nothing to commit` | No changes detected | `git status` | Make changes first, or check if files are gitignored |
| `CONFLICT` during merge | Same file modified in both branches | `git status` shows "both modified" | Edit conflicting file, resolve markers, `git add .`, `git commit` |
| `detached HEAD` warning | You checked out a specific commit, not a branch | `git status` shows "detached HEAD" | `git switch main` to return to a branch |
| `fatal: remote origin already exists` | Trying to add a remote that's already configured | `git remote -v` | Use `git remote set-url origin <url>` instead |
| Can't push | Someone else pushed first | `git status` shows behind remote | `git pull` first, then `git push` |
| Accidentally committed secrets | `.env` or API key in a commit | `git log --oneline` shows the commit | Remove file, add to .gitignore, commit, then rewrite history if pushed |

## 13. Common Mistakes

| Mistake | Why It's a Problem | How to Avoid |
|---------|-------------------|--------------|
| **Committing secrets** | API keys visible to anyone with repo access | Always add `.env` to `.gitignore` first |
| **Committing the venv** | Hundreds of MBs of unnecessary files | Check `.gitignore` before first commit |
| **Vague commit messages** | Can't find what changed or why | Follow project commit convention |
| **Not committing often** | Large, unclear changes hard to review | Commit each completed unit |
| **Committing generated files** | `__pycache__/`, `.pyc`, model artifacts | Add to `.gitignore` |
| **Force pushing to main** | Overwrites others' work | Never force push to shared branches |
| **Not pulling before push** | Merge conflicts | Always `git pull` before `git push` |
