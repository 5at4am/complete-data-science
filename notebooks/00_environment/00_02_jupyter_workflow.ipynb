## 1. What Are We Solving?

Every single notebook in this curriculum is a Jupyter notebook. If you cannot navigate cells, manage kernels, or run code reliably, you will waste hours on tooling instead of learning machine learning.

This notebook teaches you the **workflow** — not just what cells are, but how to think about notebooks as executable documents where code, explanation, and output live together.

**The problem:** Beginners treat notebooks like scripts (run top-to-bottom, never look back) or like documents (read-only). Neither works. Notebooks are **interactive scratchpads** that must also be **reproducible documents**.

---

## 2. Why Does It Matter?

| Scenario | Consequence |
|----------|-------------|
| You skip cells and variables are undefined | `NameError` — looks like a bug, actually a workflow problem |
| You restart kernel and code fails | Hidden state — your notebook depended on something you forgot |
| You use the wrong kernel | Packages not found, even though you installed them |
| You never save | Hours of work gone |
| You put 50 lines in one cell | Can't isolate which line broke |

**Real-world example:** A data scientist shares a notebook with a colleague. The colleague runs it and gets `ModuleNotFoundError`. The original author never ran "Restart & Run All" — their notebook only worked because they ran cells in a specific manual order with packages loaded in a previous session.

This is the **#1 notebook pitfall** in industry. This notebook teaches you to avoid it.

---

## 3. Mental Model

Think of a notebook as a **conversation between you and the computer**:

```text
┌─────────────────────────────────────────────────────┐
│                   NOTEBOOK                          │
│                                                     │
│  Cell 1: "Here's what I'm going to do" (markdown)   │
│  Cell 2: "Let's try it" (code) → output             │
│  Cell 3: "Here's what happened" (markdown)           │
│  Cell 4: "Let's go deeper" (code) → output           │
│                                                     │
│  ═══════════════════════════════════════════════════  │
│  KERNEL (Python process): holds ALL variables        │
│  x = 10, y = 20, plt = <module>, np = <module>      │
│  ═══════════════════════════════════════════════════  │
└─────────────────────────────────────────────────────┘
```

**Key insight:** The kernel is a **stateful** process. Variables from Cell 1 exist in Cell 5. This is powerful (you can build on previous work) but dangerous (you can accidentally depend on state you forgot about).

**Analogy:** A kernel is like a whiteboard. Everything you write stays until you erase it (restart). If you write "x = 10" on one part and reference it later, it works — but if someone erases the board and you forget, you're stuck.

## 6. Cells

A notebook is a sequence of **cells**. There are three types:

| Cell Type | Purpose | Executed? | Use For |
|-----------|---------|-----------|----------|
| **Code** | Runs Python | Yes | Experiments, calculations, plots |
| **Markdown** | Renders formatted text | No | Explanations, titles, tables |
| **Raw** | Passes content unchanged | No | Custom export formats |

### Cell Modes

A cell can be in two modes:

| Mode | Indicator | What Happens |
|------|-----------|---------------|
| **Edit** | Green border, cursor visible | You're typing content |
| **Command** | Blue border, no cursor | Keyboard shortcuts work |

Toggle between them with `Esc` (to Command) and `Enter` (to Edit).

### Essential Keyboard Shortcuts

| Shortcut | Action | When to Use |
|----------|--------|-------------|
| `Shift+Enter` | Run cell, move to next | Execute current cell |
| `Ctrl+Enter` | Run cell, stay on it | Testing a cell |
| `Alt+Enter` | Run cell, insert new below | Quick experiment |
| `A` | Insert cell above | Add context before |
| `B` | Insert cell below | Add follow-up |
| `DD` | Delete cell | Remove mistakes |
| `Z` | Undo | Recover deleted cell |
| `M` | Convert to Markdown | Add explanation |
| `Y` | Convert to Code | Make explanation executable |
| `Ctrl+Shift+-` | Split cell at cursor | Break up long cells |

### Code Cells

Code cells run Python. The output appears below the cell. The last expression is auto-displayed (no `print()` needed for quick inspection).

````python
# This is a code cell
print("Hello from a code cell!")
x = 10
y = 20
print(f"x + y = {x + y}")
````

### Markdown Cells

Markdown cells render formatted text. You can use:

- **Bold** and *italic*
- `inline code`
- Headings, lists, tables
- LaTeX math: $y = mx + b$

```python
# This is a code block in markdown (not executed)
print("This won't run")
```

> **Important:** Markdown code blocks are *displayed* but not *executed*. They're for showing code examples, not running them.

### When to Use Markdown vs Code

| Situation | Use |
|-----------|-----|
| Explaining what you're about to do | Markdown |
| Showing a command without running it | Markdown code block |
| Actually running Python | Code cell |
| Displaying results of analysis | Code cell (auto-display) or Markdown |
| A note to yourself or others | Markdown |

### Best Practice: Alternate Markdown and Code

A good notebook flows like:

```text
Markdown: "Let's load the data and look at the first 5 rows"
Code:     pd.read_csv(...) / df.head()
Markdown: "We can see the data has 3 columns..."
Code:     df.describe()
Markdown: "The mean price is $X, which suggests..."
```

This makes the notebook readable as a document, not just a script.

## 7. Kernels

The **kernel** is the Python process that executes code cells. Key points:

- Each notebook has one kernel.
- Variables persist between cells (shared state).
- **Restart the kernel** to clear all state.
- **Restart & Run All** runs the notebook from a clean state.

### The State Problem

When you run Cell 1, variables enter the kernel's memory. Cell 2 can use them. Cell 3 can use them. This continues until you restart.

```text
Kernel Memory After Running Cells:

Cell 1: x = 10        → kernel: {x: 10}
Cell 2: y = x + 5     → kernel: {x: 10, y: 15}
Cell 3: print(x + y)  → kernel: {x: 10, y: 15} → outputs 25

But what if someone runs Cell 3 first?
Cell 3: print(x + y)  → kernel: {} → NameError: x not defined
```

### Why Restart & Run All Matters

Notebooks can accidentally depend on hidden state (variables defined in earlier cells that you forgot). Running from a clean kernel catches this. **Always verify notebooks with Restart & Run All.**

| Method | What It Does | When to Use |
|--------|-------------|-------------|
| Run Cell | Executes one cell | Testing a change |
| Run All | Runs every cell top-to-bottom | After editing multiple cells |
| Restart & Run All | Clears state, then runs everything | **Before sharing, committing, or finishing** |

> **Rule of thumb:** If a notebook doesn't pass "Restart & Run All", it's not done.

## 8. Magic Commands

Magic commands start with `%` (line magic) or `%%` (cell magic). They control the notebook environment. Think of them as Jupyter's built-in shortcuts.

### Line Magic (`%`)

Applies to one line:

```python
%timeit sum(range(1000))  # times this one line
```

### Cell Magic (`%%`)

Applies to the entire cell. Must be the first line:

```python
%%writefile my_script.py
# This entire cell content is written to my_script.py
print("Hello from the file")
```

### Common Magic Commands

| Command | Type | Purpose | Example |
|---------|------|---------|---------|
| `%timeit` | Line | Time a statement | `%timeit sum(range(1000))` |
| `%matplotlib inline` | Line | Show plots inline | Must be before `import matplotlib` |
| `%load_ext` | Line | Load an extension | `%load_ext autoreload` |
| `%%writefile` | Cell | Write cell to file | `%%writefile test.py` |
| `%run` | Line | Run a Python file | `%run my_script.py` |
| `%who` | Line | List variables | `%who` |
| `%whos` | Line | List variables with types | `%whos` |
| `%reset` | Line | Clear all variables | `%reset -f` |
| `%cd` | Line | Change directory | `%cd ../data` |
| `%%time` | Cell | Time entire cell | `%%time` on first line |

### `%matplotlib inline` — Critical for This Curriculum

This magic command must appear **before** you import matplotlib. It tells Jupyter to display plots directly in the notebook. Without it, plots appear in separate windows or not at all.

```python
# CORRECT ORDER:
%matplotlib inline          # ← first
import matplotlib.pyplot as plt  # ← then import

# WRONG ORDER:
import matplotlib.pyplot as plt  # ← import first
%matplotlib inline          # ← too late, already bound
```

````python
# %timeit measures execution time
%timeit sum(range(1000))
````

````python
# %matplotlib inline makes plots appear in the notebook
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title("Inline plot")
plt.show()
````

### What to Look For: Good vs Bad Signals

| Signal | Good | Bad |
|--------|------|-----|
| Variables defined | Before they're used | Referenced but never defined |
| Plots | Appear inline | Pop up in separate windows |
| Kernel indicator | Solid circle (idle) | Filled circle (busy, still running) |
| Cell numbers | Sequential: 1, 2, 3... | Gaps: 1, 5, 12 (cells were skipped) |
| Outputs | Present and sensible | Empty or errors |
| Notebook runs clean | Restart & Run All passes | Errors on restart |

## 9. Best Practices

1. **Run from top to bottom** — never skip cells.
2. **Restart & Run All** to verify from a clean state.
3. **Use markdown** to explain what and why.
4. **Use deterministic seeds** (`np.random.seed(42)`) for reproducibility.
5. **Avoid hidden state** — don't rely on variables from cells you skipped.
6. **Keep cells focused** — one idea per cell.
7. **Explain non-obvious code** with comments.

## 10. Troubleshooting Guide

| Symptom | Cause | Verify | Fix |
|---------|-------|--------|-----|
| `NameError: name 'x' is not defined` | Variable not yet defined (cell skipped or kernel restarted) | Check cell order; run all cells from top | Run `Shift+Enter` from the top, or Restart & Run All |
| `ModuleNotFoundError` | Package not installed in active environment | `pip list` or `conda list` | `pip install <package>` or use correct kernel |
| Plot doesn't appear | `%matplotlib inline` not set, or wrong order | Check if magic is before import | Add `%matplotlib inline` before `import matplotlib` |
| Cell says `[*]` for a long time | Code is still running (infinite loop or slow computation) | Look for `while True` or heavy computation | Interrupt kernel (Restart button) |
| Output is gone | Kernel was restarted | Look for `In [*]` or empty outputs | Re-run all cells |
| Wrong Python version | Kernel points to wrong environment | `import sys; print(sys.executable)` | Change kernel via Kernel > Change kernel |
| Notebook won't save | Disk full or permissions issue | Check disk space | Free space or move to different directory |

## 11. Common Mistakes

| Mistake | Why It's a Problem | How to Avoid |
|---------|-------------------|--------------|
| **Hidden state** | Notebook works in your session but fails for others | Always use Restart & Run All before sharing |
| **Wrong kernel** | Packages not found even though they're installed | Check kernel matches your virtual environment |
| **Not saving** | Lose hours of work | Enable autosave; `Ctrl+S` frequently |
| **Giant cells** | Hard to debug which line caused the error | Split into small, focused cells |
| **Running cells out of order** | Variables have unexpected values | Use Restart & Run All to verify order-independence |
| **Forgetting `%matplotlib inline`** | Plots don't appear or appear in popup windows | Put it in the first code cell |
| **Not using markdown** | Notebook is unreadable code dump | Add markdown explanations between code cells |
