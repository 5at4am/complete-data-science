## 5. Cells

A notebook is a sequence of **cells**. There are two main types:

- **Code cells**: contain Python code, executed by the kernel.
- **Markdown cells**: contain formatted text (like this).

### Code Cells

Code cells run Python. The output appears below the cell.

````python
# This is a code cell
print("Hello from a code cell!")
x = 10
y = 20
print(f"x + y = {x + y}")
````

### Markdown Cells

Markdown cells render formatted text. You can use:

- **Bold** and *italic*
- `inline code`
- Headings, lists, tables
- LaTeX math: $y = mx + b$

```python
# This is a code block in markdown (not executed)
print("This won't run")
```

## 6. Kernels

The **kernel** is the Python process that executes code cells. Key points:

- Each notebook has one kernel.
- Variables persist between cells (shared state).
- **Restart the kernel** to clear all state.
- **Restart & Run All** runs the notebook from a clean state.

### Why Restart & Run All Matters

Notebooks can accidentally depend on hidden state (variables defined in earlier cells that you forgot). Running from a clean kernel catches this. **Always verify notebooks with Restart & Run All.**

````python
# Variables persist between cells
print(f"x is still {x} from the earlier cell")
````

## 7. Magic Commands

Magic commands start with `%` (line magic) or `%%` (cell magic). They control the notebook environment.

````python
# %timeit measures execution time
%timeit sum(range(1000))
````

````python
# %matplotlib inline makes plots appear in the notebook
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title("Inline plot")
plt.show()
````

### Common Magic Commands

| Command | Purpose |
|---------|---------|
| `%timeit` | Time a statement |
| `%matplotlib inline` | Show plots inline |
| `%load_ext` | Load an extension |
| `%%writefile` | Write cell content to a file |
| `%run` | Run a Python file |
| `%who` | List variables |
| `%whos` | List variables with details |

````python
# %whos shows current variables
%whos
````

## 8. Best Practices

1. **Run from top to bottom** — never skip cells.
2. **Restart & Run All** to verify from a clean state.
3. **Use markdown** to explain what and why.
4. **Use deterministic seeds** (`np.random.seed(42)`) for reproducibility.
5. **Avoid hidden state** — don't rely on variables from cells you skipped.
6. **Keep cells focused** — one idea per cell.
7. **Explain non-obvious code** with comments.

## 9. Common Mistakes

- **Hidden state**: notebook works because you ran cells in a specific order, but fails on Restart & Run All.
- **Wrong kernel**: using a different Python than your environment.
- **Not saving**: losing work.
- **Giant cells**: hard to debug.
